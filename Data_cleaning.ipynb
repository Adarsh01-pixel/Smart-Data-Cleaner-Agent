```python
# Step 4: Generate Sales Dataset

import pandas as pd
import numpy as np

# Set random seed
np.random.seed(42)

# Number of sales records
n = 500

# Create sales data
data = {

    "Order_ID": range(1001, 1001 + n),

    "Customer_Name": np.random.choice(
        ["Rahul", "Priya", "Amit", "Sneha", "Karan", "Neha"],
        n
    ),

    "Product": np.random.choice(
        ["Laptop", "Mobile", "Tablet", "Headphones", "Keyboard"],
        n
    ),

    "Category": np.random.choice(
        ["Electronics", "Accessories", "Gadgets"],
        n
    ),

    "Quantity": np.random.randint(1, 10, n),

    "Unit_Price": np.random.choice(
        [500, 1000, 1500, 2000, 2500, 5000],
        n
    ),

    "City": np.random.choice(
        ["Pune", "Mumbai", "Delhi", "Bangalore", "Nashik"],
        n
    ),

    "Payment_Method": np.random.choice(
        ["UPI", "Cash", "Credit Card", "Debit Card"],
        n
    ),

    "Order_Date": pd.date_range(
        start="2026-01-01",
        periods=n,
        freq="D"
    )
}

# Convert dictionary into DataFrame
df = pd.DataFrame(data)

# Calculate total sales
df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]

# Display confirmation
print("Sales Dataset Created Successfully!")
print("Total Records:", len(df))
print("Total Columns:", len(df.columns))
```

```python
recovery_check = city_cleaned.copy()

# Recover missing Quantity
mask_quantity = (
    recovery_check["Quantity"].isna()
    & recovery_check["Unit_Price"].notna()
    & recovery_check["Unit_Price"].gt(0)
)

recovery_check.loc[mask_quantity, "Quantity"] = (
    recovery_check.loc[mask_quantity, "Total_Sales"]
    / recovery_check.loc[mask_quantity, "Unit_Price"]
)

# Recover missing Unit_Price
mask_price = (
    recovery_check["Unit_Price"].isna()
    & recovery_check["Quantity"].notna()
    & recovery_check["Quantity"].gt(0)
)

recovery_check.loc[mask_price, "Unit_Price"] = (
    recovery_check.loc[mask_price, "Total_Sales"]
    / recovery_check.loc[mask_price, "Quantity"]
)

display(recovery_check[
    ["Order_ID", "Quantity", "Unit_Price", "Total_Sales"]
].head(15))
```

```python
print("Missing Quantity:", recovery_check["Quantity"].isna().sum())
print("Missing Unit_Price:", recovery_check["Unit_Price"].isna().sum())

print("Negative Quantity:", (recovery_check["Quantity"] < 0).sum())
print("Negative Unit_Price:", (recovery_check["Unit_Price"] < 0).sum())

print(
    "Sales mismatches:",
    (
        recovery_check["Total_Sales"]
        != recovery_check["Quantity"] * recovery_check["Unit_Price"]
    ).sum()
)
```

```python
invalid_records = recovery_check[
    (recovery_check["Quantity"] < 0)
    | (recovery_check["Unit_Price"] < 0)
]

display(invalid_records[
    ["Order_ID", "Product", "Quantity", "Unit_Price", "Total_Sales"]
])
```

```python
recovery_check["Review_Status"] = "Valid"

invalid_mask = (
    (recovery_check["Quantity"] < 0)
    | (recovery_check["Unit_Price"] < 0)
)

recovery_check.loc[invalid_mask, "Review_Status"] = "Manual Review Required"

display(
    recovery_check.loc[
        invalid_mask,
        ["Order_ID", "Quantity", "Unit_Price", "Total_Sales", "Review_Status"]
    ]
)
```
